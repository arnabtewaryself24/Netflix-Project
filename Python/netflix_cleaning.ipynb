```python
import pandas as pd
import numpy as np

df = pd.read_excel("../Data/Raw/Netflix TV Shows and Movies.xlsx")

df.head()
df.info()
df.shape
df.isnull().sum()

df = df.drop_duplicates()

print(df.shape)

df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)

df.columns

df["description"] = df["description"].fillna("No Description Available")
df["age_certification"] = (df["age_certification"].fillna("Unknown"))

median_votes = df["imdb_votes"].median()

df["imdb_votes"] = (df["imdb_votes"].fillna(median_votes))

df.isnull().sum()

df["runtime"] = pd.to_numeric(
    df["runtime"],
    errors="coerce"
)

df["runtime"] = (
    df["runtime"]
      .fillna(df["runtime"].median())
)

df["runtime"].describe()

df["release_year"] = pd.to_numeric(
    df["release_year"],
    errors="coerce"
)

df["release_year"] = (
    df["release_year"]
      .fillna(df["release_year"].median())
      .astype(int)
)
df.info()

df["type"].value_counts()
df["imdb_score"].describe()

df.to_csv(
    "../Data/Cleaned/netflix_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```

```python
import pandas as pd

df = pd.read_csv("../Data/Cleaned/netflix_cleaned.csv")

# Remove line breaks from all text fields
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].astype(str).str.replace(r'[\r\n]+', ' ', regex=True)

# Save a clean version
df.to_csv(
    "../Data/Cleaned/netflix_cleaned_fixed.csv",
    index=False,
    encoding="utf-8"
)

print("New file created: cleaned_fixed.csv")
```

```python
import pandas as pd
from sqlalchemy import create_engine

# Read CSV
df = pd.read_csv("../Data/Cleaned/netflix_cleaned_pg.csv")

# Connect to PostgreSQL
engine = create_engine(
    "postgresql+psycopg2://postgres:Arnab%4073649@localhost:5432/netflix_db"
)

# Load data into PostgreSQL
df.to_sql(
    "netflix_titles",
    engine,
    if_exists="replace",   # Creates/Replaces table
    index=False
)

print(f"Successfully loaded {len(df)} rows into PostgreSQL!")
```
